# MNIST exploration

```python
import sys; sys.path.append('..')
import matplotlib.pyplot as plt
from scripts.train import load_mnist
train_x,train_y,_,_=load_mnist('../data')
fig,axes=plt.subplots(2,8,figsize=(10,3))
for ax,image,label in zip(axes.flat,train_x[:16],train_y[:16]):
 ax.imshow(image.squeeze(),cmap='gray'); ax.set_title(int(label)); ax.axis('off')
```
