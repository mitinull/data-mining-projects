```python
import numpy as np, matplotlib.pyplot as plt
data=np.load('../outputs/artifacts.npz')
def show(images,title,columns=10):
 fig,axes=plt.subplots(int(np.ceil(len(images)/columns)),columns,figsize=(10,2));
 for ax,image in zip(axes.flat,images): ax.imshow(image.squeeze(),cmap='gray'); ax.axis('off')
 fig.suptitle(title); fig.tight_layout()
show(data['beta_1_samples'],'Random samples (β=1)')
```

```python
show(data['beta_1_interpolation'],'Interpolation between two latent means')
show(data['beta_1_grid'],'2-D latent grid',columns=12)
```
